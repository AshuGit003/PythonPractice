```python
#Q. Accept a number from user and insert that element at the end of tuple

elem = int(input("Enter your number: "))
tup1 = (1,2,3,4)

list1 = list(tup1)
list1.append(elem)

tup1 = tuple(list1)

print(tup1)
```

```python
t1 = (12,34,65,123,45,12,12)
print(t1.index(65))
print(t1.count(12))
print(min(t1))
print(max(t1))
print(sum(t1))
print(len(t1))

del t1
```

```python
#Packing of tuple
std1 = (101,"Ashish","C#")
std2 = (102,"Ganesh","Java")
std3 = (103,"Omkar","Python")

#UnPacking of tuple
RollNo,Name,Language = std1
print(RollNo,Name,Language)
```

```python
#Nested tuple=> Tuple inside another tuple
students = (
    (101,"Ashish","C#"),
    (102,"Ganesh","Java"),
    (103,"Omkar","Python")
)

print(students[0][1])
```

```python
#4. Write a Python program that removes an item from a tuple by creating a new tuple.
#   The original tuple should remain unchanged.
t1 = (3,4,5,6)
t2 = (4,5)

s1 = set(t1)
s2 = set(t2)

s3 = s1 - s2
t3 = tuple(s3)
print(t3)
```

```python
#5. Write a Python program that finds and prints the length of the longest string in a tuple of strings.
str_tup = ('ssd','asds','asdffdf','asdssd','ffvfseewfef')

max = 0
for i in str_tup:
    if len(i) > max:
        max = len(i)
        max_str = i

print(f"{max_str} has maximum length of {max}")
```
