Tuples

```python
#Q. Accept a number from user and insert that element at the end of tuple

elem = int(input("Enter your number: "))
tup1 = (1,2,3,4)

list1 = list(tup1)
list1.append(elem)

tup1 = tuple(list1)

print(tup1)
```

```python
t1 = (12,34,65,123,45,12,12)
print(t1.index(65))
print(t1.count(12))
print(min(t1))
print(max(t1))
print(sum(t1))
print(len(t1))

del t1
```

```python
#Packing of tuple
std1 = (101,"Ashish","C#")
std2 = (102,"Ganesh","Java")
std3 = (103,"Omkar","Python")

#UnPacking of tuple
RollNo,Name,Language = std1
print(RollNo,Name,Language)
```

```python
#Nested tuple=> Tuple inside another tuple
students = (
    (101,"Ashish","C#"),
    (102,"Ganesh","Java"),
    (103,"Omkar","Python")
)

print(students[0][1])
```
