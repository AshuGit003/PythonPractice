```python
# Q1. Write Data to a File
# Write a Python program that:
# 1. Takes the student's name, age, course, and city as input.
# 2. Stores the information in student.txt.
# 3. Displays "Student data saved successfully."

try:
    name = str(input("Enter Name: "))
    age = int(input("Enter Age: "))
    course = str(input("Enter Course: "))
    city = str(input("Enter City: "))
    with open("student.txt","w") as file:
        file.write(f"Name: {name}, Age: {age}, Course: {course}, City: {city}")
except Exception as e:
    print(e)
else:
    print("Student data saved successfully.")
```

```python
# Q2. Read Data from a File
# Write a program to read student.txt and display all its contents on the screen.

with open("student.txt","r") as file:
    output = file.read()
    print(output)
```

```python
# Q3. Write Multiple Lines
# Create a file named college.txt and store the following information:
# College Name
# Department
# Course
# Academic Year
# Read the file and display each line separately.


try:
    CollegeName = str(input("Enter College Name: "))
    Department = str(input("Enter Department: "))
    Course = str(input("Enter Course: "))
    AcademicYear = int(input("Enter Academic Year: "))

    with open("college.txt","w") as file:
        file.write(f"\n College Name: {CollegeName} \n Department: {Department} \n Course: {Course} \n Academic Year: {AcademicYear}")

    with open("college.txt","r") as file:
        for line in file.readlines():
            print(line.strip("\n"))
except Exception as e:
    print(e)
```

```python
# Q4. Append Data
# Create students.txt containing one student's information.
# Write a program that accepts another student's:
# - Name
# - Course
# - Marks
# and appends the information to the existing file without deleting previous data.

name = str(input("Enter Name: "))
course = str(input("Enter Course: "))
marks = int(input("Enter Marks: "))

with open("student.txt","a") as file:
    file.write(f"\nName: {name}, Course: {course}, Marks: {marks}")

with open("student.txt","r") as file:
    for line in file.readlines():
        print(line.strip("\n"))
```

```python
# Q5. Count Lines
# Write a Python program to read students.txt and display:
# Total number of students: ?
# Assume that each student is stored on a separate line.

with open("student.txt","r") as file:
    count = 0
    for line in file.readlines():
        count += 1

    print(f"Total numbers of students are {count}")
```
