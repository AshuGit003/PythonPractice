```python
main_str=input("Enter Your Main String: ")
sub_str=input("Enter Your Substring: ")

if sub_str in main_str:
    print(f"{sub_str} exist in {main_str}.")
else:
    print(f"{sub_str} does not exist in {main_str}.")
```

```python
str1=input("Enter String One: ")
str2=input("Enter String Two: ")

if sorted(str1) == sorted(str2):
    print("Strings are anagram.")
else:
    print("Strings are not anagram.")
```

```python
import re

def myCode(pattern, string):
    match = re.search(pattern,string)
    if match:
        print(f"Pattern '{pattern}' found!") 
    else:
        print(f"Pattern '{pattern}' not found in the string.")   
    

pattern = "abc"
string = "abcdef abc abcghijkl"
myCode(pattern,string)


```
