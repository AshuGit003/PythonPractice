Sorting Algorithm In Python

```python
# Bubble Sort 
# Swapping the adjecent elements if they are in the wrong order

list1 = [64,34,25,12,22,11,90]

for i in range(len(list1)):
    for j in range(len(list1)):
        if list1[i] < list1[j]:
            temp = list1[i]
            list1[i] = list1[j]
            list1[j] = temp

print(list1)
```

```python
# Selection Sort
# Comparison-based sorting algorithm. 
# It sorts an array by repeatedly selecting the smallest (or largest) element from the unsorted portion 
# and swapping it with the first unsorted element.

list1 = [64,25,12,22,11,34,90]

n = len(list1)
for i in range(n):

    min_idx = i
    for j in range(i+1,n):
        if list1[j] < list1[min_idx]:
            min_idx = j

    list1[i], list1[min_idx] = list1[min_idx], list1[i]

print(list1) 

```

```python
# Insertion Sort
# Iteratively inserting each element of an unsorted list into its correct position in a sorted portion of the list

list1 = [12,11,13,5,6]
#[11,12,13,5,6]
#[5,11,12,13,6]

n = len(list1)
for i in range(1,n):
    key = list1[i]      #6
    j = i - 1           #[3]

    while j >= 0 and list1[j] > key:   # 5 > 6
        list1[j+1] = list1[j]           # [5,11,11,12,13]
        j -= 1                         # [0]

    list1[j+1] = key

print(list1) 
```
