Python Regex

```python
import re

text = "I am learning Python"
result = re.search("Python",text)

if result:
    print("Match found !!!")
else:
    print("Match not found !!!")
```

```python
import re

text = "I am learning Python"
result = re.search(r"Python$",text)

if result:
    print("Ends with Python !!!")
else:
    print("Not ends with Python !!!")
```

```python
import re

text = "I am learning Python"
result = re.search(r"^Python",text)

if result:
    print("Starts with Python !!!")
else:
    print("Not starts with Python !!!")
```

```python
import re

text = "I am learning5 Python"
result = re.search(r"\d",text)

if result:
    print("Digit present in the given text !!!")
else:
    print("Digit not present in the given text !!!")
```

```python
import re

text = "I am learning Python"
result = re.search(r"[qsw]",text)

if result:
    print("q,s or w present in the given text !!!")
else:
    print("q,s or w not present in the given text !!!")
```

```python
import re

text = str(input("Enter some text: "))
result = re.search(r"[A-Z]",text)

if result:
    print("Upper case is present")
else:
    print("Upper case is not present.")
```

```python
import re

text = str(input("Enter some text: "))
result = re.search(r"^\d",text)

if result:
    print("Starts with digit.")
else:
    print("Does not start with digit.")
```

```python
#Check email id valid or not
email = str(input("Enter your email ID: "))

result = re.search(r"^[a-zA-Z]+[0-9]+[@]+[a-zA-Z]+[.]+[a-zA-Z]", email)

if result:
    print("Valid email")
else:
    print("Invalid email")
```
