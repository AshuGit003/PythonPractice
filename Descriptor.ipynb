Python Descriptor

```python
class Student:
    def getAge(self):
        print("Getter is called!!!")
        return self._age

    def setAge(self,age):
        print("Setter is called!!!")
        self._age = age

    def delAge(self):
        print("Deletter is called!!!")
        del self._age

    age = property(getAge, setAge, delAge)

s = Student()
s.age = 25
print(s.age)
s.delAge()
```

```python
#Descriptor using annotation

class Person:
    @property
    def age(self):
        print("Getter is called!!!")
        return self._age

    @age.setter
    def age(self, age):
        if 18 <= age <= 100:   
            self._age = age
        else:
            print("Not Eligible for voting")
            self._age = None

    @age.deleter
    def age(self):
        print("Deleter is called!!!")
        del self._age

s = Person()
s.age = 13
print(s.age)
del s.age    
```

```python
# Create a descriptor name to validate name of a person.
# 1.Name should be string
# 2.Length of name should not be more than 8.

class Person:
    @property
    def name(self):
        return self._name

    @name.setter
    def name(self,name):
        if isinstance(name,str) and len(name) <= 8:
            self._name = name
        else:
            print("Invalid Name")
            self._name = None

    @name.deleter
    def name(self):
        del self._name

p = Person()
p.name = "Ashish"
print(p.name)
del p.name
```
