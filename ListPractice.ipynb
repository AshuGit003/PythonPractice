```python
#7.Write a Python program that finds the kth largest element in a list of numbers without sorting the list.
numList = [23,63,86,25,75,15,97,76]

def find_kth_largest_manual(nums, k):
    # Store the first k elements as our tracking list
    largest_elements = nums[:k]
    
    # Sort only this small tracking list initially
    largest_elements.sort(reverse=True)
    
    # Check the rest of the elements in the main list
    for num in nums[k:]:
        # If the current number is larger than the smallest in our tracking list
        if num > largest_elements[-1]:
            largest_elements[-1] = num
            largest_elements.sort(reverse=True)
            
    return largest_elements[-1]

print(find_kth_largest_manual(numList,2))
```
