Python Numpy

NumPy is a Python package. It stands for 'Numerical Python'. It is a library consisting of multidimensional array objects and a collection of routines for processing of array.

```python
import numpy as np
```

```python
a = np.array([1,2,3])
print(a)
print(type(a))
```

```python
a = np.array([23,43,45,65,34],dtype=str)
print(a)
print(type(a))
print(a[2])
```

```python
# 2-D Array

a = np.array([[23,43,45],[65,73,35]])
b = np.array([[8,9,4],[5,2,6],[7,2,5]])

print(a)
print(b)

print(a[1][1])
print(b[0][2])
print(a.shape)
print(b.shape)
print(a.shape[0])
print(a.shape[1])
```

```python
np.array([1,2.0,3,4]) #Here, NumPy has upcast integer values to float values
```

```python
a = np.array([(1,2,3,4),(5,6,7,8)],dtype=int)

print(a)
print("Shape of Array b:",a.shape)
print("Number of rows:",a.shape[0])
print("Number of columns:",a.shape[1])
```

```python
a1=np.array([12,123,3,4])
a1.shape
```

```python
c=np.array([(1,2,3,4),(7,8,9,10),(17,18,19,20)],dtype=int)
print(c)
print(c[1][1])
print(c[2][2])
```

```python
# Array of zeros

a = np.zeros(4, dtype=int)
b = np.zeros((4,3))

print(a)
print(b)
```

```python
np.zeros((2,4))+1100
```

```python
# Array of ones

np.ones(5)
```

```python
np.ones((3,5))
```

```python
# Random numbers in ndarrays

a = np.random.rand(3,4)

print(a)
```

```python
np.round(12.345456,decimals=3)
```

```python
# round(array,decimal_size)
a = np.round(np.random.rand(3,3),decimals=3)
print(a)
```

```python
#np.random.randint(start,end,(row_size,col_size))
a = np.random.randint(10,30,(3,4))
print(a)
```

```python
# An array of your choice
a = np.full((3,3),5)
print(a)
```

```python
#Imatrix in Numpy
a = np.eye(3)
print(a)
```

```python
b = np.eye(5,k=2)
print(b)
```

```python
# Evenly Spaced ndarray
range(1,10,2)
```

```python
np.arange(2,15,1)
```

```python
np.arange(0.5,10.0,0.5) #it accepts float values as well
```

```python
# linspace()

print(np.linspace(10,20,8))
```

```python
# number of axis
a = np.array([[2,5,7],[8,6,1]])
print("Array:",'\n',a)
print('Dimensions :','\n',a.ndim)

b = np.array([5,10,15])
print('Array :','\n',b)
print('Dimensions :','\n',b.ndim)
```
