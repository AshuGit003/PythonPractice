```python
set2 = {34,445,76,23,665,23,12,65}
set2.add(43)
set2.update([79,37])
set2.pop()
set2.remove(34) #To remove specific element => Raise an error if element is not present
set2.discard(900) #To remove specific element
print(set2)
```

```python
#Operations
A = {1,2,3}
B = {3,4,5}

#Using operators
print("Union: ",A|B)
print("Intersection: ",A & B)
print("Difference: ", A - B)
print("Difference: ", B-A)
print("Symmetric Difference: ", A^B)

#Using functions
print("Union: ",A.union(B))
print("Intersection: ",A.intersection(B))
print("Difference: ", A.difference(B))
print("Symmetric Difference: ", A.symmetric_difference(B))

```

```python
s1 = {1,2,3,4,5}
s2 = {1,2,3}

print(s1.issuperset(s2))
print(s2.issubset(s1))
print(s1.isdisjoint(s2)) # Returns True IF both sets dont have any common element
```
