```python
#Operations
A = {1,2,3}
B = {3,4,5}

#Using operators
print("Union: ",A|B)
print("Intersection: ",A & B)
print("Difference: ", A - B)
print("Difference: ", B-A)
print("Symmetric Difference: ", A^B)

#Using functions
print("Union: ",A.union(B))
print("Intersection: ",A.intersection(B))
print("Difference: ", A.difference(B))
print("Symmetric Difference: ", A.symmetric_difference(B))

```

```python
s1 = {1,2,3,4,5}
s2 = {1,2,3}

print(s1.issuperset(s2))
print(s2.issubset(s1))
print(s1.isdisjoint(s2)) # Returns True IF both sets dont have any common element
```

```python
#Frozenset => Immutable version of set
X = frozenset([2,3,4,5])
print(X)
```

```python
#Membership operator => IN :It checks whether the element is present in given set or not without using if statemnet
Y = {3,4,5,6}
print(3 in Y)
print(9 in Y)
```

```python
from itertools import product

list1 = [1, 2, 2, 3]
list2 = [3, 4, 4, 5]

# Convert to sets to get unique elements first
unique_l1 = set(list1)  # {1, 2, 3}
unique_l2 = set(list2)  # {3, 4, 5}

unique_pair = set(product(unique_l1,unique_l2))
print(unique_pair)

result = []
for i in unique_l1:
    for j in unique_l2:
        result.append((i, j))

print(result)
```
