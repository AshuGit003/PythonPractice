Python Exception Handling

```python
try:
    a =int(input("Enter a: ")) 
    b =int(input("Enter b: "))
    print(a/b)
except:
    print("Can not devided by zero !!!") 
```

```python
try:
    a =int(input("Enter a: ")) 
    b =int(input("Enter b: "))
    print(a/b)
except ZeroDivisionError:
    print("Can not devided by zero !!!")
except ValueError:
    print("Please Enter Numeric Value Only !!!")
else:
    print("Else block")
```

```python
try:
    a =int(input("Enter a: ")) 
    b =int(input("Enter b: "))
    print(a/b)
except Exception as e:
    print(e)
else:
    print("Else block in exception handling !!!")
finally:
    print("Finally block in exception handling !!!")
```

```python
try:
    with open("Student.txt","w") as file:
        file.write("Hii I'm Ashish !!!")

    with open("Student22.txt","r") as file:
        output = file.read()
        print(output)
except Exception as e:
    print(e)
else:
    print("no error")
finally:
    file.close()

```
