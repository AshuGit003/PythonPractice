Python Exception Handling

```python
try:
    a =int(input("Enter a: ")) 
    b =int(input("Enter b: "))
    print(a/b)
except:
    print("Can not devided by zero !!!") 
```

```python
try:
    a =int(input("Enter a: ")) 
    b =int(input("Enter b: "))
    print(a/b)
except ZeroDivisionError:
    print("Can not devided by zero !!!")
except ValueError:
    print("Please Enter Numeric Value Only !!!")
else:
    print("Else block")
```

```python
try:
    a =int(input("Enter a: ")) 
    b =int(input("Enter b: "))
    print(a/b)
except Exception as e:
    print(e)
else:
    print("Else block in exception handling !!!")
finally:
    print("Finally block in exception handling !!!")
```

```python
try:
    with open("Student.txt","w") as file:
        file.write("Hii I'm Ashish !!!")

    with open("Student22.txt","r") as file:
        output = file.read()
        print(output)
except Exception as e:
    print(e)
else:
    print("no error")
finally:
    file.close()

```

```python
#Raise => telling which error is occured
try:
    roll = int(input("Enter your roll number: "))
    if roll <= 0:
        raise ValueError()
    print("Roll no:" ,roll)
except ValueError:
    print("ValueError Exception thrown")
```

```python
# Validate user for eligible for Voting

try:
    age = int(input("Enter age: "))
    if age < 1 or age > 100:
        raise ValueError("Invalid age !!!")
    elif age >= 18:
        print("Eligible to vote !!!")
    else:
        print("Not eligible to vote !!!")
except ValueError as e:
    print(e)
```

```python
#Write a program to accept number from user. If input is not a number throw exception
#"Please enter number only!!!". Ask for input number from user until user enter a number.

while True:
    try:
        num = int(input("Enter number: "))
        print(f"Entered number is {num}")
        break
    except ValueError:
        print("Please enter number only !!!")
```

```python
#defining a new exception named MyCustomError
class MyCustomError(Exception):
    pass
roll = int(input("Please enter your roll number"))
if roll <= 0:
    raise MyCustomError("The number entered is not positive")
```

```python
#4. Write a Python program that opens a file data.txt in write mode,
# write some content, and ensures the file is closed properly by
# using the finally block

try:
    file = open("xyz.txt","w+")
    file.write("Welcome to exception handling/nWelcome to exception logging")
    file.close()
    file = open("xyz.txt","r")
    file.seek(0)
    data = file.read()
    len_data = len(data)
    word_pos = len_data - 7
    file.seek(word_pos)
    word = file.read()
    print(word)
    print(data)

except FileNotFoundError:
    print("File Not Found")
finally:
    file.close() 
```
