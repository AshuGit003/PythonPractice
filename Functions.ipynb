Python Functions

```python
#Positional Argument

def student(name,age):
    print(name, age)

student("Ashish","24")
```

```python
#Keyword Arguments

def employee(name, salary):
    print(name, salary)

employee(salary="1000000", name = "Ashish")
```

```python
# Default Argument

def country(name = "India"):
    print(name)

country()
country("Australia")
```

```python
#Variable Length 
# *args=> Used to pass multiple values (non-keyword argument)

def total(*numbers):
    print(sum(numbers))

total(10,20,30)

#**kwargs => Used to pass multiple keyword arguments

def details(**data):
    print(data)

details(name="Ashish", city="Pune")
```

```python
#1.

def greet():
    print("Hello, World!")

greet()
```

```python
#2.
def sum_of_two(a, b):
    return a + b

print(sum_of_two(5,7))
```

```python
#3.
def multiply(a, b=2):
    return a*b

print(multiply(10,20))
print(multiply(4)) 
```

```python
#4.
def get_square(num):
    return num **2

print(get_square(5))
```
