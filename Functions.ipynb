Python Functions

```python
#Positional Argument

def student(name,age):
    print(name, age)

student("Ashish","24")
```

```python
#Keyword Arguments

def employee(name, salary):
    print(name, salary)

employee(salary="1000000", name = "Ashish")
```

```python
# Default Argument

def country(name = "India"):
    print(name)

country()
country("Australia")
```

```python
#Variable Length 
# *args=> Used to pass multiple values (non-keyword argument)

def total(*numbers):
    print(sum(numbers))

total(10,20,30)

#**kwargs => Used to pass multiple keyword arguments

def details(**data):
    print(data)

details(name="Ashish", city="Pune")
```

```python
#1. Write a function greet() that prints "Hello, World!" when called. Call the function.

def greet():
    print("Hello, World!")

greet()
```

```python
#2. Define a function sum_of_two(a, b) that takes two parameters a and b and returns their sum. 
# Call the function with the values 5 and 7.
def sum_of_two(a, b):
    return a + b

print(sum_of_two(5,7))
```

```python
#3. Define a function multiply(a, b=2) that takes two parameters a and b, where b defaults to 2 if not provided.
# Call the function with both one and two arguments.
def multiply(a, b=2):
    return a*b

print(multiply(10,20))
print(multiply(4)) 
```

```python
#4. Define a function get_square(num) that returns the square of a given number num. Call the function and print the result.
def get_square(num):
    return num **2

print(get_square(5))
```

```python
#5. Define a function concatenate_strings(str1, str2) that concatenates two given strings and returns the result.
# Call the function with "Hello" and "World".

def concatenate_strings(str1, str2):
    return str1 + " " + str2

print(concatenate_strings("Hello","World"))
```

```python
#6. Define a function describe_pet(animal_type, pet_name) that accepts animal_type and pet_name as keyword arguments. 
# It should print a message like "I have a pet dog named Rover.".
# Call the function with appropriate keyword arguments.

def describe_pet(animal_type, pet_name):
    print(f"I have a pet {animal_type} named {pet_name}.")

describe_pet(animal_type="Cat",pet_name="Lio")
```

```python
#7. Write a function sum_all(*args) that sums up any number of integer arguments passed to it. 
# Call the function with a list of numbers.

def sum_all(*args):
    sum = 0
    for i in args:
        sum += i

    return sum

print(sum_all(1,2,3))
```

```python
#8. Write a function min_max(nums) that takes a list of numbers and returns both the minimum and maximum numbers as a tuple.
# Call the function with a list of integers.

def min_max(*nums):
    min_num = min(nums)
    max_num = max(nums)
    return min_num,max_num

print(min_max(20,30,40))
```
