```python
#1.
num = int(input("Enter a number: "))

count = 0
for i in range(2,num + 1):
    if num % i == 0:
        count += 1

if count == 1:
    print(f"{num} is Prime number.")
else:
    print(f"{num} is not Prime number.")
```

```python
#2.
primeList = []

for num in range(50,101):
    count = 0
    for i in range(2,num + 1):
        if num % i == 0:
            count += 1
        if count == 2:
            break

    if count == 1:
        primeList.append(num)

print("Prime List: ",primeList)
```

```python
#3.
num = int(input("Enter a number: "))
reversedNum = 0

while num > 0:
    lastNum = num % 10
    reversedNum = (reversedNum * 10) + lastNum
    num = num // 10

print(reversedNum)
    
```

```python
#4.
num = int(input("Enter a number: "))
n = num
reversedNum = 0

while n > 0:
    lastNum = n % 10
    reversedNum = (reversedNum * 10) + lastNum
    n = n // 10

if num == reversedNum:
    print(f"{num} is Palindrome number")
else:
    print(f"{num} is not Palindrome number")
```

```python
#5.
num = int(input("Enter a number: "))
sum = 0

while num > 0:
    lastNum = num % 10
    sum += lastNum
    num = num // 10

print(sum)
```

```python
#6.
num = int(input("Enter a num: "))
n = num
pow = 4

sum = 0

while n > 0:
    lastNum = n % 10
    sum += (lastNum**pow)
    n = n // 10

if sum == num:
    print(f"{num} is armstrong number")
else:
    print(f"{num} is not armstrong number")
```
