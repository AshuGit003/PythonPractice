Dictionary

```python
Employee = {
    "EmpNo":101,
    "Name":"Ashish", 
    "Location":["Pune","Mumbai","Banglore"],
    "Salary":650000,
    "Basic Salary":650000
    }

print(Employee)
```

```python
Bike= dict(
    Brand="TVS",
    Model="Raider",
    Price=120000,
    Milage="55 km/ltr",
    Color=["Black","Red","Blue"]
)

print(Bike)
```

```python
#1. 
print(Employee["EmpNo"])
print(Bike["Model"])

#2.
print(Employee.get("Name"))
print(Bike.get("Price"))
```

```python
#Adding element in dict
#1.
Employee["Phone"] = 788787789
Bike["Fuel_type"] = "Petrol"

#2.
Employee.update({"Email":"abc@123"})
Bike.update({"Owner":"Ashish"})

print(Employee)
print(Bike)
```

```python
#deleting an element
#1. pop()
Employee.pop("Email")
print(Employee)
```

```python
#2. popitem()
Bike.popitem()
print(Bike)
```

```python
#3. del
del Employee["Basic Salary"]
print(Employee)
```

```python
#4. Clear
Bike.clear()
print(Bike)
```

```python
#Copieng a dictionary
#1. copy()
new_employee = Employee.copy()
print(new_employee)

#2. dict()
new_bike = dict(Bike)
print(new_bike)
```

```python
myFamily = {
    "child1": {
        "name": "Akash",
        "age": 21,
        "DOB": {
            "date":23,
            "month":7,
            "year": 1999 
        },
        "study": "2nd year"
    },

    "child2": {
        "name": "Rushi",
        "age": 22,
        "DOB": {
            "date":13,
            "month":6,
            "year": 1998 
        },
        "study": "3rd year",
        "marks":{
            "UT1":45,
            "UT2":89,
            "UT3":97
        }
    },

    "child3":{
        "name": "Naveen",
        "age": 24
    }
}

print(myFamily)
```

```python
#access
print(myFamily["child2"]["DOB"]["year"])
```

```python
#update
myFamily["child2"]["marks"]["UT3"] = 95
print(myFamily)
```

```python
#delete
del myFamily["child2"]["marks"]["UT3"]
print(myFamily)
```

```python
print(Employee.keys())
print(Employee.values())
print(Employee.items())
```

```python
for i in Employee.items():
    print(i)
```

```python
for key in Employee.keys():
    print(key)
```

```python
for value in Employee.values():
    print(value)
```

```python
#dictionary comprehension
Student = {
    "EmpNo":101,
    "Name":"Ashish", 
    "Location":"Pune"
    }
student = {key:value for key,value in Student.items()}
print(student)
```

```python
numbers = [1,2,3,4,5,6]
num_cube = {num:num **3 for num in numbers if num % 2 == 1}
print(num_cube)
```
