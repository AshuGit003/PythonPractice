Dictionary

```python
Employee = {
    "EmpNo":101,
    "Name":"Ashish", 
    "Location":["Pune","Mumbai","Banglore"],
    "Salary":650000,
    "Basic Salary":650000
    }

print(Employee)
```

```python
Bike= dict(
    Brand="TVS",
    Model="Raider",
    Price=120000,
    Milage="55 km/ltr",
    Color=["Black","Red","Blue"]
)

print(Bike)
```

```python
#1. 
print(Employee["EmpNo"])
print(Bike["Model"])

#2.
print(Employee.get("Name"))
print(Bike.get("Price"))
```

```python
#Adding element in dict
#1.
Employee["Phone"] = 788787789
Bike["Fuel_type"] = "Petrol"

#2.
Employee.update({"Email":"abc@123"})
Bike.update({"Owner":"Ashish"})

print(Employee)
print(Bike)
```
