```python
#dictionary comprehension
Student = {
    "EmpNo":101,
    "Name":"Ashish", 
    "Location":"Pune"
    }
student = {value:key for key,value in Student.items()}
print(student)
```

```python
#1. Write a program that iterates through the dictionary and prints both the keys and values.
person = {"name": "John", "age": 25, "city": "New York"}

for i in person.items():
    print(i)
```

```python
#2. Write a Python program that finds the key with the maximum value in a dictionary.

numbers = {"Key1":34,"Key2":76,"Key3":29,"Key4":89,"Key5":62}

value_list = numbers.values()
max_value = max(value_list)

for key in numbers.items():
    if key[1] == max_value:
        print(key[0])
```

```python
#3. Write a program that creates a dictionary from a list of numbers, where the keys are the numbers, 
# and the values are their squares.
numbers = [1,2,3,4,5]

dict = {}
for i in numbers:
    dict[i] = i **2

print(dict)
```

```python
#4. Write a Python program that inverts a dictionary, i.e., swaps keys with values.
dict = {"Sun":1,"Mon":2,"Tue":3,"Wed":4,"Thu":5,"Fri":6,"Sat":7}

dict = {val:key for key,val in dict.items()}
print(dict)
```

```python
#6. 
employees = {"John": 50000, "Alice": 60000, "Bob": 45000,"Diana": 70000,"Charlie": 55000}

# 1. Print all employee names.
employeeName = list(emp for emp in employees.keys())
print(employeeName)

# 2. Print all salaries.
salaries = list(salary for salary in employees.values())
print(salaries)

# 3. Calculate the total salary of all employees.
sum = 0
for sal in employees.values():
    sum += sal
print(sum)

# 4. Calculate the average salary.
avg = sum//len(employees)
print(avg)

# 5. Add a new employee "Eva": 65000 to the dictionary.
employees["Eva"] = 65000
print(employees)

# 6. Update "Bob"’s salary to 48000.
employees["Bob"] = 48000
print(employees)

# 7. Remove "Charlie" from the dictionary.
employees.pop("Charlie")
print(employees)

# 8. Check if "Alice" exists in the dictionary.
print("Alice" in employees.keys())

# 9. Create a new dictionary of employees whose salary is more than 55000.
result = {name: salary for name, salary in employees.items() if salary > 55000}
print(result)

# 10. Count how many employees earn less than 60000.
result = {salary for salary in employees.values() if salary < 60000}
print(len(result))
```
