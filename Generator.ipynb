Python Generator Function

Generator-Function: A generator-function is defined like a normal function, but whenever it needs to generate a value, it does so with a yeild keyword rather than return. If the body of def contains yield, the function automatically becomes a generator function.

```python
def Demo():
    yield 10
    yield 20
    yield 30
```

```python
a = Demo()
```

```python
next(a)
```

```python
#Convert iterable into iterator/generator
def genFun(lst1):
    for i in lst1:
        yield i
```

```python
list1 = [123,345,7,65857,98,456,63,4]
g = genFun(list1)
```

```python
next(g)
```

```python
#Convert iterable in to iterator by changing brackets of list comprehension
list1 = [123,345,7,65857,98,456,63,4]
g = (i*2 for i in list1)
```

```python
next(g)
```

```python
#Using iter()
list1 = [123,345,7,65857,98,456,63,4]
g=iter(list1)
```

```python
next(g)
```

```python
# Write a generator function Gen_seq() to generate sequence of numbers from 101 to 200.

def Gen_seq():
    for i in range(101,201):
        yield i

g = Gen_seq()
```

```python
next(g)
```

```python
def Gen_seq(start,end):
    i = start

    while i < end +1:
        yield i
        i += 1

start_index = int(input("Enter Start Index: "))
end_index = int(input("Enter End Index: "))

g = Gen_seq(start_index,end_index)
```

```python
next(g)
```
