```python
# 1.Create a class ATM with methods: check_balance(), deposit(), and withdraw().
# Create an object and perform different transactions.

class ATM:
    def __init__(self):
        self.balance = 2000

    def check_balance(self):
        print(f"Current Balance: {self.balance}")

    def deposit(self):
        self.amount = int(input("Enter Amount to Deposit: "))
        self.balance += self.amount
        print(f"Amount deposited. Your current balance is {self.balance}")

    def withdraw(self):
        self.amount = int(input("Enter Amount to Withdraw: "))
        self.balance -= self.amount
        print(f"Amount withdrawn. Your current balance is {self.balance}")

atm = ATM()

op = int(input("Choose Option:\n 1.Check Balance\n 2.Deposit\n 3.Withdraw"))

if op == 1:
    atm.check_balance()
elif op == 2:
    atm.deposit()
elif op == 3:
    atm.withdraw()
else:
    print("invalid Input")
```

```python
#2.Create a Python program using Class and Object to manage student results.
# Create a class named Student with the following methods:
# setData() – Accept student name and marks of 3 subjects.
# calculate() – Calculate total marks and percentage.
# display() – Display student details, total, percentage and grade.
# Grade Criteria:
# Percentage ≥ 75 → Grade A
# Percentage ≥ 60 → Grade B
# Percentage ≥ 50 → Grade C
# Below 50 → Fail
# Requirements:
# Create 2 student objects.
# Store different data for both students.
# Display the result of both students.
# Challenge: Display which student has the highest percentage.

class Student:
    def setData(self, name, marksList):
        self.name = name
        self.marksList = marksList
        self.totalMarks = 0
        self.percentage = 0.0
        self.grade = ""

    def calculate(self):
        self.totalMarks = sum(self.marksList)
        self.percentage = (self.totalMarks/300) * 100

        if self.percentage >= 75:
            self.grade = "Grade A"
        elif self.percentage >= 60:
            self.grade = "Grade B"
        elif self.percentage >= 50:
            self.grade = "Grade C"
        else:
            self.grade = "Fail"

    def display(self):
        print(f"Name: {self.name}, Total Marks: {self.totalMarks}, Percentage: {self.percentage}, Grade: {self.grade}")

s1 = Student()
s2 = Student()

s1.setData("Gauri",[97,78,89])
s2.setData("Ashish",[78,65,67])

s1.calculate()
s2.calculate()

s1.display()
s2.display()

if s1.percentage > s2.percentage:
    print(f"{s1.name} has highest percentage !!!")
else:
    print(f"{s2.name} has highest percentage !!!")
```

```python
#1.Animal and Dog
#Create an Animal class with a eat() method. Create a Dog child class with a bark() method. 
# Create an object and call both methods.

class Animal:
    def eat(self):
        print("Can eat")

class Dog(Animal):
    def bark(self):
        print("Can bark")

d1 = Dog()
d1.eat()
d1.bark()
```

```python
#2.Person and Teacher
# Create a Person class with a constructor that accepts name and age.
# Create a Teacher child class with a teach() method. Display the person's details and teaching message.

class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def display_details(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")

class Teacher(Person):
    def teach(self):
        print("Can teach")

t1 = Teacher("Ashish",24)
t1.display_details()
t1.teach()
```

```python
#3.BankAccount and SavingsAccount
# Create a BankAccount parent class with account holder name and balance. 
# Create a SavingsAccount child class with an additional interest-rate attribute. 
# Use super() to initialize the parent attributes.

class BankAccount:
    def __init__(self, ach_name, balance):
        self.ach_name = ach_name
        self.balance = balance

class SavingAccount(BankAccount):
    def __init__(self, ach_name, balance, interest):
        super().__init__(ach_name, balance)
        self.interest = interest

    def display_details(self):
        print(f"Account Holder Name: {self.ach_name}")
        print(f"Balance: {self.balance}")
        print(f"Interest Rate: {self.interest}")

s1 = SavingAccount("Ashish",10000000,8)
s1.display_details()
```

Multiple Inheritance

```python
#Q1. Person → Student → Result
# Create three classes:
# - Person: Store the student's name.
# - Student: Store roll number.
# - Result: Store marks.
# Create an object of Result and display the name, roll number, and marks.

class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, rollNo):
        super().__init__(name)
        self.rollNo = rollNo

class Result(Student):
    def __init__(self, name, rollNo, marks):
        super().__init__(name, rollNo)
        self.marks = marks

    def display_details(self):
        print(f"Name: {self.name}")
        print(f"Roll No: {self.rollNo}")
        print(f"Marks: {self.marks}")

re = Result("Ashish",21,45)
re.display_details()
```

```python
#Q8. Hospital Management System
# Create Person → Patient → InPatient.
# - Person: Name and age
# - Patient: Patient ID and disease
# - InPatient: Room number and number of days admitted
# Calculate the total room charge if the daily charge is ₹2,000.

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Patient(Person):
    def __init__(self, name, age, patientId, disease):
        super().__init__(name, age)
        self.patientId = patientId
        self.disease = disease

class InPatient(Patient):
    def __init__(self, name, age, patientId, disease, roomNumber, admittedDays):
        super().__init__(name, age, patientId, disease)
        self.roomNumber = roomNumber
        self.admittedDays = admittedDays

    def calculate_roomCharge(self):
        charge = self.admittedDays * 2000
        return charge

    def display_details(self):
        print(f"Patient Name: {self.name}")
        print(f"Patient Age: {self.age}")
        print(f"Patient ID: {self.patientId}")
        print(f"Disease: {self.disease}")
        print(f"Room Number: {self.roomNumber}")
        print(f"Admitted Days: {self.admittedDays}")
        print(f"Total Charges: {self.calculate_roomCharge()}")

p1 = InPatient("XYZ",78,101,"Cancer",121,4)
p1.display_details()
```

Polymorphism

```python
class Animal:
    def sound(self):
        print("Animal make sound")

class Dog(Animal):
    def sound(self):
        print("Dog say woof")

class Cat(Animal):
    def sound(self):
        print("Cat say Meow")

dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

```python
class Shape:
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

class Rectangle(Shape):
    def __init__(self,l,b):
        self.l = l
        self.b = b

    def area(self):
        return self.l * self.b

class Triangle(Shape):
    def __init__(self,b,h):
        self.b = b
        self.h = h

    def area(self):
        return 0.5 * self.b * self.h

c1 = Circle(5)
print(c1.area())

r1 = Rectangle(4,3)
print(r1.area())

t1 = Triangle(5,6)
print(t1.area())
```

```python
class CreditCard:
    def pay(self, amount):
        print(f"Paid {amount} using Credit Card")

class UPI:
    def pay(self, amount):
        print(f"Paid {amount} using UPI")

class Cash:
    def pay(self, amount):
        print(f"Paid {amount} using Cash")

def make_payment(payment_method, amount):
    payment_method.pay(amount)

card = CreditCard()
upi = UPI()
cash = Cash()

make_payment(card, 1000)
make_payment(upi, 2000)
make_payment(cash,3000)
```
